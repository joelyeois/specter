```python
import matplotlib.pyplot as plt

from cryosim.detectors import falcon4i_300kv, k3_300kv, perfect_detector

%load_ext autoreload
%autoreload 2
```

```python
dx = 1.0
n = 256
k_f, falcon4i = falcon4i_300kv(n, dx, return1d=True)
k, k3 = k3_300kv(n, dx, return1d=True)
k_p, perfect = perfect_detector(n, dx, return1d=True)
k_pp, pperfect = perfect_detector(512, 0.5, return1d=True)
plt.plot(k_f, falcon4i, label="Falcon 4i")
plt.plot(k, k3, label="Gatan K3")
plt.plot(k_p, perfect, ls="--", label="Perfect")
plt.plot(k_pp, pperfect, ls="--", label="Perfect x2 Super-res")
plt.ylim([0, 1])
plt.xlim([0, 0.5])
plt.xlabel("k (1/A)")
plt.ylabel("MTF")
plt.legend()
# plt.imshow(mtf, vmin=0, vmax=1, cmap='bone')
# plt.colorbar()
```
